```python
class AlexNet(nn.Module):
    def __init__(self, iamge_size: int = 64):
        super().__init__()
        self.net = nn.Sequential(
            # nn.Conv2d(3, 9, 3),
            # nn.BatchNorm2d(9),
            # nn.MaxPool2d(2),
            # nn.ReLU(),
            # nn.Conv2d(9, 12, 3),
            # nn.BatchNorm2d(12),
            # nn.MaxPool2d(2),
            # nn.ReLU(),

            # nn.Conv2d(12, 18, 3),
            # nn.ReLU(),
            # nn.Conv2d(18,18, 3),
            # nn.ReLU(),
            # nn.Conv2d(18,12, 3),
            # nn.BatchNorm2d(12),
            # nn.ReLU(),
            # nn.MaxPool2d(2),

            # nn.Flatten(),
            # nn.Linear(192, 400),
            # nn.ReLU(),
            # # nn.Dropout(),
            # nn.Linear(400, 400),
            # nn.ReLU(),
            # nn.Dropout(),
            # nn.Linear(400, 200)
              nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Flatten(),
            nn.Linear(128 * 8 * 8, 200),
       )

    def forward(self, x):
        return self.net(x)
```

```python
model = AlexNet(64)
model
```

```python
# train
optim =torch.optim.AdamW(model.parameters(), lr = 1e-3)
```

```python
def to_3ch(img):
    if img.ndim == 2:  # H x W
        img = img.unsqueeze(0)
    if img.shape[0] == 1:
        img = img.repeat(3, 1, 1)
    return img
```

```python
epochs = 10000
batch_size = 32
for epoch in range(epochs):
    # ix = torch.randint(0, tiny_imagenet.shape[0], (8,),).tolist()
    # # ix = 
    # # xb, yb = trans_data[ix]["image"], trans_data[ix]["label"]
    # # xb = xb.float()/ 255.0
    # # yb = yb.long()
    # xb = torch.stack(trans_data[ix]["image"]).float() / 255.0
    # yb = torch.tensor(trans_data[ix]["label"]).long()

    ix = torch.randint(0, len(tiny_imagenet), (batch_size,)).tolist()
    images = [to_3ch(trans_data[i]["image"]) for i in ix]
    xb = torch.stack(images).float() / 255.0
    yb = torch.tensor([trans_data[i]["label"] for i in ix]).long()


    pred = model(xb)

    optim.zero_grad()
    loss = nn.functional.cross_entropy(pred, yb)
    loss.backward()
    optim.step()
    if epoch % 500 == 0:
        print(f"step {epoch}, loss: {loss.item()}")
    
```
